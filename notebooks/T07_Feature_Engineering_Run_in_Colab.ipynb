## 2. Put files in the project folders
The source CSV belongs in `data/processed/` and the Python functions belong in `src/`.

```python
from pathlib import Path
import shutil

Path("src").mkdir(exist_ok=True)
Path("data/processed").mkdir(parents=True, exist_ok=True)

for filename, destination in [
    ("features.py", "src/features.py"),
    ("responses_clean.csv", "data/processed/responses_clean.csv"),
]:
    if not Path(filename).exists():
        raise FileNotFoundError(f"Please upload {filename} in step 1.")
    shutil.move(filename, destination)
    print("Ready:", destination)
```
