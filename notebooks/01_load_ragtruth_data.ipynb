# Load RAGTruth data

This notebook loads the raw RAGTruth JSONL files and confirms the expected row counts.

```python
from pathlib import Path

import pandas as pd

repo_root = Path.cwd()
if not (repo_root / 'data' / 'raw').exists():
    repo_root = repo_root.parent

data_dir = repo_root / 'data' / 'raw'

response_path = data_dir / 'response.jsonl'
source_path = data_dir / 'source_info.jsonl'

response_df = pd.read_json(response_path, lines=True)
source_df = pd.read_json(source_path, lines=True)

print(f'response rows: {len(response_df)}')
print(f'source rows: {len(source_df)}')

response_df.head()
source_df.head()
```
